```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

Save squared values of FFT coefficients (Training data)

```python
input_dir = Path("img200x200")
output_dir = Path("pow_dft200x200")
output_dir.mkdir(exist_ok=True)

last_pow_dft = None
last_image_index = None

for i in range(1, 11):
    img_path = input_dir / f"image_{i:05d}.jpg"

    img = Image.open(img_path).convert("L")
    arr = np.asarray(img, dtype=np.float64)

    # 2D Discrete Fourier Transform
    dft_arr = np.fft.fft2(arr)

    # Shift the zero-frequency component to the center
    dft_shifted = np.fft.fftshift(dft_arr)

    # Power spectrum (Fourier-plane intensity)
    pow_dft = np.abs(dft_shifted) ** 2

    # Save the power spectrum
    np.save(output_dir / f"pow_dft_{i:05d}.npy", pow_dft)

    # Save the last sample for display
    last_pow_dft = pow_dft
    last_image_index = i

print(f"Saved files to: {output_dir.resolve()}")

# ---------------------------------------------------------
# Display the last Fourier-plane image
# ---------------------------------------------------------

plt.figure(figsize=(6, 6))

plt.imshow(
    np.log1p(last_pow_dft),    # logarithmic display
    cmap="gray",
    origin="lower",
    interpolation="nearest"
)

plt.title(f"Fourier Plane Intensity (image_{last_image_index:05d}.jpg)")
plt.xlabel("Spatial Frequency $f_x$")
plt.ylabel("Spatial Frequency $f_y$")
plt.colorbar(label=r"$\log(1+|F|^2)$")
plt.tight_layout()
plt.show()
```
